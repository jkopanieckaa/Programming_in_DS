### Zadanie 1
Napisz funkcję, która przyjmie dwa argumenty: przychód i koszt. Funkcja zwraca TRUE/True jeśli zysk jest dodatni, w przeciwnym razie FALSE/False.


```python
def czy_zysk(przychod, koszt):
    return przychod - koszt > 0

print(czy_zysk(5000,4000))
```

### Zadanie 2
Napisz funkcję, która przyjmuje listę kwot i sprawdza, czy wszystkie wartości są większe od zera. Podpowiedź: użyj funkcji all().

```python
kwoty = [2000,300,4000,-5]

def wieksze_od_zera(kwoty):
    return all(x > 0 for x in kwoty)

print(wieksze_od_zera(kwoty))
```

Napisz funkcję, która dla podanego procentu inflacji i nominalnej stopy procentowej sprawdza, czy realna stopa procentowa jest większa niż 0.

```python
procent_inflacji = 0.2
nominalna_stopa_procentowa = 0.1

def realna_stopa_wikesza_od_zera(inflacja,nominalna_stopa):
    return nominalna_stopa - inflacja > 0

print(realna_stopa_wikesza_od_zera(procent_inflacji, nominalna_stopa_procentowa))
```

### Zadanie 4
Napisz funkcję, która zwraca kwotę netto po potrąceniu podatku (23%). Jeśli kwota przekracza 1000, zwróć po potrąceniu, jeśli nie, zwróć kwotę bez zmian.


```python
def kwota_netto(kwota):
    if kwota > 1000:
        podatek = kwota * 0.23
        return kwota - podatek
    else:
        return kwota

print(kwota_netto(900))
print(kwota_netto(9000))
```

### Zadanie 5
Napisz funkcję, która dla dwóch argumentów typu liczba zwraca większą z nich, ale tylko jeśli obie są różne i dodatnie; w przeciwnym razie zwraca None (Python) / NA (R).

```python
def ktora_wieksza(a,b):
    if a > 0 and b > 0 and a != b:
        if a > b:
            return a
        else:
            return b

#mozna użyc max(a,b) wtedy wyświrtla większą liczbe zamiast kolejnego if

def wieksza_liczba(a, b):
    if a > 0 and b > 0 and a != b:
        return max(a, b)



print(ktora_wieksza(9000,9000))
print(ktora_wieksza(9000,1000))
print(ktora_wieksza(-1000,9000))

print(wieksza_liczba(900,-2))
```

### Zadanie 6
Napisz funkcję, która przyjmuje trzy liczby i zwraca TRUE/True, jeśli co najmniej dwie z nich są większe od 5000.

```python
def co_najmniej_dwie_wieksze_od_5000(a,b,c):
    return sum(x > 5000 for x in [a,b,c]) >= 2

print(co_najmniej_dwie_wieksze_od_5000(9000,800,10))
```

### Zadanie 7
Napisz funkcje , która dla przekazanej listy lub wektora liczb:

obliczają średnią i odchylenie standardowe,

zwracaja te wyniki:
jako krotkę
jako słownik

Następnie, w każdym wariancie, wyświetl oba wyniki łącznie oraz z osobna.

```python
import statistics

lista = [3000,2000,5000]
def obliczenia_listy_wektora_krotka(x):
    srednia = round(statistics.mean(x),2)
    odchylenie_standardowe = round(statistics.stdev(x),2)

    return srednia, odchylenie_standardowe

def obliczenia_listy_wektora_slownik(x):
    srednia = round(statistics.mean(x),2)
    odchylenie_standardowe = round(statistics.stdev(x),2)

    return {"Średnia": srednia, "Odchylenie standardowe": odchylenie_standardowe}

wynik_krotka = obliczenia_listy_wektora_krotka(lista)
print(wynik_krotka)
print(wynik_krotka[0])
print(wynik_krotka[1])

wynik_slownik = obliczenia_listy_wektora_slownik(lista)
print(wynik_slownik)
print(wynik_slownik.get("Średnia"))
print(wynik_slownik.get("Odchylenie standardowe"))

```
