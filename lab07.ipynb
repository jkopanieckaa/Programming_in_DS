### Zadanie 5
Policz sumę wartości większych od średniej w wektorze 20 losowych liczb zmiennoprzecinkowych.

```python
vec = np.random.rand(20)
suma = vec[vec > np.mean(vec)].sum()
print(suma)
```

### Zadanie 6
Dla tablicy 1D z 12 liczb: przypisz 0 do wszystkich liczb mniejszych od 5, a 1 do pozostałych.

```python
tab = np.arange(12)
new_tab = [0 if i < 5 else 1 for i in tab]
# lub
# new_tab = np.where(tab < 5,0,1)
print(new_tab)
```

### Zadanie 7
Mając dwie tablice 1D, stwórz trzecią zawierającą maksimum z elementów o danym indeksie.

### Zadanie 8
Od każdego elementu wektora 11:20 odejmij numer jego indeksu; zastosuj wektoryzację.

```python
vec = np.arange(11,21)
wynik = vec - [i for i in range(10)]
print(vec)
print(wynik)
```

### Zadanie 9
Oblicz iloczyn skalarny dwóch losowych wektorów długości 6. podpowiedź: w Pythonie użyj funkcji np.dot(), w R użyj funkcji sum() wraz z mnożeniem wektorów.

```python
vec_a = np.random.rand(6)
vec_b = np.random.rand(6)

wynik = np.dot(vec_a, vec_b)
print(vec_a)
print(vec_b)
print('\n',wynik)
```

### Zadanie 10
Dla macierzy 2x5 z losowymi wartościami od 0 do 100, zamień elementy nieparzyste na 0 (zachowaj wektoryzację).

```python
a = np.random.randint(0,101,10).reshape(2,5)
wynik = np.where(a % 2 != 0,0,a)
#a[a%2 ==1] = 0

print(wynik)
```
