### Zadanie 1

Oblicz miesięczną ratę kredytu, znając kapitał, liczbę miesięcy i oprocentowanie roczne.

#### rata = kapital / liczba_miesiecy * (1 + oprocentowanie)


```python
kapital = 1000
liczba_miesiecy = 12
oprocentowanie_roczne = 0.09

rata = kapital / liczba_miesiecy * (1 + oprocentowanie_roczne)

print(rata)
print(round(rata,2))
```

### Zadanie 3

Oblicz procent wzrostu wartości inwestycji między dwoma latami.


```python
wartosc_poczatkowa = 1000
wartosc_koncowa = 2050

procent_wzrostu = (wartosc_koncowa - wartosc_poczatkowa) / wartosc_poczatkowa * 100

print(procent_wzrostu)
```

Użyj pętli while, aby zasymulować spadek ceny produktu o 10% co rok, aż cena będzie mniejsza niż 50.
