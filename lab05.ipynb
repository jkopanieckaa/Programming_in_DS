### Zadanie 1
Napisz funkcję, która dla podanej listy miesięcznych przychodów i kosztów zwraca listę zysków miesięcznych (przychód minus koszt dla każdego miesiąca). Wykorzystaj pętlę for.

```python
przychody = [1000,2000,3000,4000]
koszty = [250,300,500,800]

def zyski(przychody, koszty):
    zyski = []
    for p,k in zip(przychody, koszty):
        z = p - k
        zyski.append(z)
    return zyski

print(zyski(przychody, koszty))
```

### Zadanie 2
Napisz funkcję, która dla zadanej listy cen produktów zwraca nową listę cen po podwyżce o dany procent (podany jako argument). Zastosuj pętlę for i zaokrąglaj ceny do dwóch miejsc po przecinku.

```python
ceny = [20,50,80,90]
procent = 0.2

def nowa_cena(ceny,procent_podwyżki):
    nowe_ceny = []
    for c in ceny:
        podwyzka = c * procent_podwyżki
        nowe_ceny.append(round(c + podwyzka,2))
    return nowe_ceny

print(nowa_cena(ceny, procent))
```

### Zadanie 3
Napisz funkcję, która przyjmuje liczbę i zwraca ciąg znaków oznaczający kategorię podatkową: „niska” (do 5000), „średnia” (do 15000) lub „wysoka” (powyżej 15000).

```python
def kategoria_podatkowa(x):
    if x < 5000:
        return "niska"
    elif x <= 15000:
        return "średnia"
    else:
        return "wysoka"

print(kategoria_podatkowa(1000))
print(kategoria_podatkowa(12000))
print(kategoria_podatkowa(20000))
```

### Zadanie 4
Napisz funkcję, która dla zadanej liczby kwot sprawdzi, ile z nich jest większych lub równych 1000 oraz ile mniejszych niż 1000. Zwróć oba wyniki jako dwuelementowy wektor/listę.

```python
kwoty = [200,500,700,1000,2000,1500,3500]

def kategoria_liczba(kwoty):
    mniejsze = 0
    wieksze = 0

    for k in kwoty:
        if k < 1000:
            mniejsze += 1
        elif k >= 1000:
            wieksze += 1

    return [mniejsze, wieksze]

print(kategoria_liczba(kwoty))
```

### Zadanie 5
Napisz funkcję, która dla danej listy przychodów zwraca True/TRUE, jeśli co najmniej trzy z nich są większe od średniej wszystkich przychodów.

```python
przychody = [1000,2000,3000,4000]
przychody_2 = [1,2,100,2,500,3,4]

def min_trzy_powyzej_mean(przychody):
    if len(przychody) < 3:
        return "Lista za krótka"

    srednia_przychodow = sum(przychody)/len(przychody)
    count = 0

    for p in przychody:
        if p > srednia_przychodow:
            count += 1

    return count >= 3

print(min_trzy_powyzej_mean(przychody))
print(min_trzy_powyzej_mean(przychody_2))
```
