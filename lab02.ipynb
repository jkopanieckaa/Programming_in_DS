### Zadanie 8
Utwórz ramkę danych z kolumną dat zawierającą dni od 2025-01-01 do 2025-01-10. Podpowiedź - w Pythonie użyj pd.date_range, w R seq.Date.

```python
date = pd.date_range(start="2025-01-01",end="2025-01-10",freq="D")

# dates = pd.date_range("2025-01-01", periods=10)

df = pd.DataFrame({
    "date": date
})

df
```
